```python
import pandas as pd
import os
DATA_DIRECTORY = "/content/Credify_Data"

for fname in (os.listdir(DATA_DIRECTORY)):
  if fname.endswith(".csv") and "description" not in fname:
    df = pd.read_csv(f"{DATA_DIRECTORY}/{fname}", nrows=50)
    id_of_cols = [c for c in df.columns if "SK_ID" in c]
    print(f"{fname:32s} | cols: {len(df.columns):3d} | IDs: {id_of_cols}")
```

```python
#THIS IS THE FIRST ITERATION OF TRAINING application_train.csv
import pandas as pd
train_sample = pd.read_csv("/content/Credify_Data/application_train.csv")
display(train_sample.head(50))

missing_values = train_sample.isnull().sum() / len(train_sample) * 100
missing_df = missing_values[missing_values > 0].sort_values(ascending=False)
print(f"The column of missing values is {len(missing_df)} out of {train_sample.shape[1]}")
print(missing_df.head(20))
```

```python
import numpy as np
train_clean = train_sample.copy()

num_cols = [
    c
    for c in train_clean.select_dtypes(include=np.number).columns
      if c not in ["SK_ID_CURR", "TARGET"]
]

corr_matrix = train_clean[num_cols].corr().abs()
upper = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)
drop_cols = [col for col in upper.columns if any(upper[col] > 0.95)]

print(f"Found {len(drop_cols)} columns with correlation > 0.95:")
for col in drop_cols:
  partner = upper.index[upper[col] > 0.95].tolist()
  print(f"  - Drop {col} (correlated > 0.95 with {partner})")

train_clean = train_clean.drop(columns=drop_cols)
print(f"\nShape before: {train_sample.shape} -> Shape after: {train_clean.shape}")
```

```python
# 1. Replace the 365243 placeholder anomaly (~1000 years) with NaN
train_clean["DAYS_EMPLOYED"] = train_clean["DAYS_EMPLOYED"].replace(
    365243, np.nan
)

# 2. Convert text ('object') columns into 'category' type for LightGBM
cat_cols = train_clean.select_dtypes(include=["object"]).columns.tolist()
for c in cat_cols:
  train_clean[c] = train_clean[c].astype("category")

print(f"Converted {len(cat_cols)} categorical columns.")
print(
    "DAYS_EMPLOYED max after fix:",
    train_clean["DAYS_EMPLOYED"].max(),
    "| min:",
    train_clean["DAYS_EMPLOYED"].min(),
)
```

```python
import lightgbm as lgb
import numpy as np
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import StratifiedKFold

# 1. Separate features (X) from target (y) and ID
X = train_clean.drop(columns=["SK_ID_CURR", "TARGET"])
y = train_clean["TARGET"]

# 2. Set up 5-Fold Stratified Cross-Validation
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_preds = np.zeros(len(train_clean))  # Stores out-of-fold predictions

# 3. Train across all 5 folds
for fold, (train_idx, val_idx) in enumerate(skf.split(X, y), 1):
  X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
  X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]

  model = lgb.LGBMClassifier(
      n_estimators=1000,
      learning_rate=0.05,
      random_state=42,
      n_jobs=-1,
      verbose=-1,
  )

  model.fit(
      X_train,
      y_train,
      eval_set=[(X_val, y_val)],
      callbacks=[lgb.early_stopping(stopping_rounds=50, verbose=False)],
  )

  # Predict probability of default (class 1)
  oof_preds[val_idx] = model.predict_proba(X_val)[:, 1]
  fold_auc = roc_auc_score(y_val, oof_preds[val_idx])
  print(f"Fold {fold} ROC-AUC: {fold_auc:.5f}")

# 4. Overall Cross-Validation Score
overall_auc = roc_auc_score(y, oof_preds)
print(f"\nBaseline 5-Fold OOF ROC-AUC: {overall_auc:.5f}")
```

```python
import pandas as pd

importance = pd.Series(model.feature_importances_, index=X.columns).sort_values(
    ascending=False
)

print(f"THE 50 MOST IMPORTANT FEATURES:")
print(importance.head(50))

print(f"\nFEATURES WITH ZERO IMPORTANCE:")
print(importance[importance == 0])
```

```python
# 1. Collect all feature names that had 0 splits across the baseline model
zero_imp_cols = importance[importance == 0].index.tolist()

# 2. Create a slimmed copy of train_clean without those 14 zero-importance columns
train_pruned = train_clean.drop(columns=zero_imp_cols)

print(f"Dropped {len(zero_imp_cols)} zero-importance columns:")
print(zero_imp_cols)
print(
    f"\nShape before: {train_clean.shape} -> Shape after: {train_pruned.shape}"
)
```

```python
import lightgbm as lgb
import numpy as np
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import StratifiedKFold

# 1. Separate features (X) from target (y) and ID
X_pruned = train_pruned.drop(columns=["SK_ID_CURR", "TARGET"])
y_pruned = train_pruned["TARGET"]

# 2. Set up 5-Fold Stratified Cross-Validation
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_preds_pruned = np.zeros(len(train_pruned))  # Stores out-of-fold predictions

# 3. Train across all 5 folds
for fold, (train_idx, val_idx) in enumerate(skf.split(X_pruned, y_pruned), 1):
  X_train, y_train = X_pruned.iloc[train_idx], y_pruned.iloc[train_idx]
  X_val, y_val = X_pruned.iloc[val_idx], y_pruned.iloc[val_idx]

  model_pruned = lgb.LGBMClassifier(
      n_estimators=1000,
      learning_rate=0.05,
      random_state=42,
      n_jobs=-1,
      verbose=-1,
  )

  model_pruned.fit(
      X_train,
      y_train,
      eval_set=[(X_val, y_val)],
      callbacks=[lgb.early_stopping(stopping_rounds=50, verbose=False)],
  )

  # Predict probability of default (class 1)
  oof_preds_pruned[val_idx] = model_pruned.predict_proba(X_val)[:, 1]
  fold_auc = roc_auc_score(y_val, oof_preds_pruned[val_idx])
  print(f"Fold {fold} ROC-AUC: {fold_auc:.5f}")

# 4. Overall Cross-Validation Score
overall_pruned_auc = roc_auc_score(y_pruned, oof_preds_pruned)
print(f"\nPruned (76 cols) 5-Fold OOF ROC-AUC: {overall_pruned_auc:.5f}")
print(f"Change vs baseline (0.75498):      {overall_pruned_auc - 0.75498:+.5f}")
```

```python
# Create a feature-engineered copy so train_pruned stays untouched
train_fe = train_pruned.copy()

# 1. Financial burden ratios
train_fe["CREDIT_TO_INCOME"] = (
    train_fe["AMT_CREDIT"] / train_fe["AMT_INCOME_TOTAL"]
)
train_fe["ANNUITY_TO_INCOME"] = (
    train_fe["AMT_ANNUITY"] / train_fe["AMT_INCOME_TOTAL"]
)

# 2. Employment-to-age ratio (both are negative day counts -> positive ratio)
train_fe["EMPLOYED_TO_AGE_RATIO"] = (
    train_fe["DAYS_EMPLOYED"] / train_fe["DAYS_BIRTH"]
)

# 3. Average external credit bureau score across the 3 EXT_SOURCE columns
ext_cols = ["EXT_SOURCE_1", "EXT_SOURCE_2", "EXT_SOURCE_3"]
train_fe["EXT_SOURCE_MEAN"] = train_fe[ext_cols].mean(axis=1)

print(
    f"Added 4 domain features. Shape: {train_pruned.shape} -> {train_fe.shape}"
)
```

```python
import lightgbm as lgb
import numpy as np
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import StratifiedKFold

# 1. Separate features (X_fe) from target (y_fe)
X_fe = train_fe.drop(columns=["SK_ID_CURR", "TARGET"])
y_fe = train_fe["TARGET"]

# 2. Same 5-Fold split (random_state=42) for an exact apples-to-apples comparison
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_preds_fe = np.zeros(len(train_fe))

# 3. Train across all 5 folds
for fold, (train_idx, val_idx) in enumerate(skf.split(X_fe, y_fe), 1):
  X_train, y_train = X_fe.iloc[train_idx], y_fe.iloc[train_idx]
  X_val, y_val = X_fe.iloc[val_idx], y_fe.iloc[val_idx]

  model_fe = lgb.LGBMClassifier(
      n_estimators=1000,
      learning_rate=0.05,
      random_state=42,
      n_jobs=-1,
      verbose=-1,
  )

  model_fe.fit(
      X_train,
      y_train,
      eval_set=[(X_val, y_val)],
      callbacks=[lgb.early_stopping(stopping_rounds=50, verbose=False)],
  )

  oof_preds_fe[val_idx] = model_fe.predict_proba(X_val)[:, 1]
  fold_auc = roc_auc_score(y_val, oof_preds_fe[val_idx])
  print(f"Fold {fold} ROC-AUC: {fold_auc:.5f}")

# 4. Compare against our pruned benchmark (0.75500)
overall_fe_auc = roc_auc_score(y_fe, oof_preds_fe)
print(f"\nFeature-Engineered (80 cols) 5-Fold OOF ROC-AUC: {overall_fe_auc:.5f}")
print(
    f"Change vs pruned (0.75500):                      {overall_fe_auc - 0.75500:+.5f}"
)
```

```python
import numpy as np
import pandas as pd

# 1. Align CODE_GENDER ('XNA' -> NaN) on train_fe
train_fe["CODE_GENDER"] = (
    train_fe["CODE_GENDER"].replace("XNA", np.nan).cat.remove_unused_categories()
)

# 2. Load bureau.csv
bureau = pd.read_csv("/content/Credify_Data/bureau.csv")
print("bureau.csv shape:", bureau.shape)
display(bureau.head(5))

# 3. Quick audit of types, missing %, and unique counts
display(
    pd.DataFrame({
        "dtype": bureau.dtypes,
        "missing_%": (bureau.isnull().mean() * 100).round(1),
        "nunique": bureau.nunique(),
    })
)
print("\nCREDIT_ACTIVE breakdown:\n", bureau["CREDIT_ACTIVE"].value_counts())
```

```python
# 1. Create binary indicator for currently active loans
bureau["IS_ACTIVE"] = (bureau["CREDIT_ACTIVE"] == "Active").astype(int)

# 2. Aggregate external credit history down to 1 row per SK_ID_CURR
bureau_agg = (
    bureau.groupby("SK_ID_CURR")
    .agg(
        BUREAU_LOAN_COUNT=("SK_ID_BUREAU", "count"),
        BUREAU_ACTIVE_COUNT=("IS_ACTIVE", "sum"),
        BUREAU_ACTIVE_RATIO=("IS_ACTIVE", "mean"),
        BUREAU_DAYS_CREDIT_MEAN=("DAYS_CREDIT", "mean"),
        BUREAU_DAYS_CREDIT_MAX=("DAYS_CREDIT", "max"),
        BUREAU_OVERDUE_DAYS_MAX=("CREDIT_DAY_OVERDUE", "max"),
        BUREAU_OVERDUE_AMT_SUM=("AMT_CREDIT_SUM_OVERDUE", "sum"),
        BUREAU_CREDIT_SUM=("AMT_CREDIT_SUM", "sum"),
        BUREAU_DEBT_SUM=("AMT_CREDIT_SUM_DEBT", "sum"),
        BUREAU_PROLONG_SUM=("CNT_CREDIT_PROLONG", "sum"),
    )
    .reset_index()
)

# 3. Engineer external debt-to-credit utilization ratio (+1 avoids division by zero)
bureau_agg["BUREAU_DEBT_CREDIT_RATIO"] = bureau_agg["BUREAU_DEBT_SUM"] / (
    bureau_agg["BUREAU_CREDIT_SUM"] + 1
)

# 4. Left-merge onto train_fe (keeps all 307,511 rows intact)
train_bureau = train_fe.merge(bureau_agg, on="SK_ID_CURR", how="left")

print(f"Shape before merge: {train_fe.shape}")
print(f"Shape after merge:  {train_bureau.shape}")
```

```python
import lightgbm as lgb
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import StratifiedKFold

# 1. Separate features (X_bureau) and target (y_bureau)
X_bureau = train_bureau.drop(columns=["SK_ID_CURR", "TARGET"])
y_bureau = train_bureau["TARGET"]

# 2. Exact same 5-fold split as previous benchmarks
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_preds_bureau = np.zeros(len(train_bureau))

# 3. Train across all 5 folds
for fold, (train_idx, val_idx) in enumerate(skf.split(X_bureau, y_bureau), 1):
  X_train, y_train = X_bureau.iloc[train_idx], y_bureau.iloc[train_idx]
  X_val, y_val = X_bureau.iloc[val_idx], y_bureau.iloc[val_idx]

  model_bureau = lgb.LGBMClassifier(
      n_estimators=1000,
      learning_rate=0.05,
      random_state=42,
      n_jobs=-1,
      verbose=-1,
  )

  model_bureau.fit(
      X_train,
      y_train,
      eval_set=[(X_val, y_val)],
      callbacks=[lgb.early_stopping(stopping_rounds=50, verbose=False)],
  )

  oof_preds_bureau[val_idx] = model_bureau.predict_proba(X_val)[:, 1]
  fold_auc = roc_auc_score(y_val, oof_preds_bureau[val_idx])
  print(f"Fold {fold} ROC-AUC: {fold_auc:.5f}")

# 4. Compare against previous benchmark (0.75511)
overall_bureau_auc = roc_auc_score(y_bureau, oof_preds_bureau)
print(
    f"\nWith Bureau Features (91 cols) 5-Fold OOF ROC-AUC:"
    f" {overall_bureau_auc:.5f}"
)
print(
    "Change vs train_fe (0.75511):                      "
    f" {overall_bureau_auc - 0.75511:+.5f}"
)

# 5. Check how the new BUREAU_ features ranked
imp_bureau = pd.Series(
    model_bureau.feature_importances_, index=X_bureau.columns
).sort_values(ascending=False)
print("\nTOP 15 FEATURES OVERALL:")
print(imp_bureau.head(15))
```
